# Reproducing the figures — nonlinearity-preserving quantum simulation

Each section reproduces one figure using the original, figure-specific Python plotter and the **unmodified data** for [Zenodo DOI 10.5281/zenodo.24505927](https://doi.org/10.5281/zenodo.24505927). Captions below are taken verbatim from the current `CAPTION.md` files. They reproduce the manuscript captions: reference numbers (superscript or bracketed), equation and table numbers, and Supplementary Note references refer to the accompanying paper and Supplementary Information ([arXiv:2611.23196](https://arxiv.org/abs/2611.23196)). Symbols such as `u_(st)`, `φ_(x)`, `θ_(c)` and `x_(0)` retain the plain-text subscript notation used in those caption files; model and design labels retain their definitions in the paper.

**Abbreviations:** EFTQC, early fault-tolerant quantum computing; FEM, finite element method; RMS, root mean square; POD, proper orthogonal decomposition; SINDy, sparse identification of nonlinear dynamics; KvN, Koopman–von Neumann; BE, block encoding; LCHS, linear combination of Hamiltonian simulations.

**Scope:** this notebook redraws figures from stored results. It does not rerun the flow solver, quantum transport simulations or resource-design searches. Use the `Source.Data_Production.end_to_end` workflow for numerical recalculation.

**Static artwork:** Fig. 2 has a bundled PNG preview used by `demo.ipynb`, but no numerical plotter. Supplementary Figs. S2, S6 and S7 have no public plotter or artwork. The explicitly cited numerical Source Data for Fig. S7 is included in the supplied dataset.

Run this notebook from the GitHub repository root. The first data access uses a sibling `Source_Data/` directory, `KVN_SOURCE_DATA`, a local DOI cache, or downloads `Source_Data.zip` from the published Zenodo record. Publication of the reserved DOI is required for remote downloads. Some figures need optional system tools such as Inkscape.

Plotters and the shared figure runner are in `Figure_Plots/`. Rendered files are written to `Source_Data/Generated_Outputs/Figures/<figure>/`. Numerical producers and execution drivers are in `Source/`, `Comparisons/`, and `Figure_Plots/`.


```python
from pathlib import Path
import subprocess
import sys
from IPython.display import display, Image

CODE_ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'Source/data_locations.py').is_file()), None)
if CODE_ROOT is None:
    raise RuntimeError('Open this notebook from the GitHub repository root.')
sys.path.insert(0, str(CODE_ROOT))
from Source.data_locations import source_data_root
print('Source Data:', source_data_root())

def reproduce(command, directory):
    subprocess.run([sys.executable, *command], cwd=CODE_ROOT, check=True)
    target = source_data_root() / 'Generated_Outputs/Figures' / directory
    candidates = list(target.glob('*.png'))
    if candidates:
        display(Image(filename=str(sorted(candidates)[0])))
    else:
        print('Finished; inspect output directory:', target)
```
